```python
import pandas as pd
import numpy as np
```

```python
file_path = r"C:\project placement\supply_chain_analysis\file raw\DataCoSupplyChainDataset.csv"
try:
    df = pd.read_csv(file_path, encoding='utf-8')
except UnicodeDecodeError:
    df = pd.read_csv(file_path, encoding='ISO-8859-1')
```

```python
df.info()
```

```python
#droping the unused and empty column
col_to_drop=["Order Zipcode","Product Description","Customer Email","Customer Password","Product Image","Product Status"]
df=df.drop(columns=col_to_drop)
```

```python
#filling the last name where value is na
df["Customer Lname"]=df["Customer Lname"].fillna("")
```

```python
#generating a new column with full name adding fname and lname together also after creating drop the fname and lname 
df["customer_full_name"]=df["Customer Fname"]+" "+df["Customer Lname"].str.strip()
df=df.drop(columns=["Customer Fname","Customer Lname"])
```

```python
#filling the column where zipcode is empty
df["Customer Zipcode"]=df["Customer Zipcode"].fillna(0)
```

```python
#converting str type to datetime 
df["order_date"]=pd.to_datetime(df["order date (DateOrders)"],errors="coerce")
df["shipping_date"]=pd.to_datetime(df["shipping date (DateOrders)"],errors="coerce")
df=df.drop(columns=["order date (DateOrders)","shipping date (DateOrders)"])
```

```python
df["order_year"]=df.order_date.dt.year
df["order_month"]=df.order_date.dt.month
df["order_day"]=df.order_date.dt.day_name()
df["order_hour"]=df.order_date.dt.hour
```

```python
df['delivery_delay_days'] = (df['Days for shipping (real)'] - df['Days for shipment (scheduled)'])
df['is_delayed'] = np.where(df['delivery_delay_days'] > 0, 1, 0)
```

```python
df['profit_margin_%'] = np.where(df['Sales'] > 0,
    round((df['Order Profit Per Order'] / df['Sales']) * 100, 2), 0,)
```

```python
customer_order_counts = (df.groupby('Customer Id')['Order Id'].nunique().to_dict())
#it calculate the customer unique orders and combine in dict. with customer_id
df['customer_total_orders'] = df['Customer Id'].map(customer_order_counts)
#it takes the customer id from original table and check if it exists in customer order count and then store it to customer_total_orders
```

```python
df.columns=df.columns.str.strip().str.lower().str.replace(" ","_").str.replace("(","").str.replace(")","")
df.columns
df.info()
```

```python
df.to_csv(r"C:\project placement\supply_chain_analysis\cleaned_file\supply_chain_clean.csv",index=False)
```
